```python
### 5.2. Diseño del agente (ficha PEAS)

Percepción (S):Temperatura actual del ambiente en grados Celsius.
Acciones (A): Encender aire acondicionado, encender calefacción o mantener apagado.
Entorno (E): El ambiente físico cerrado (habitación) donde está el sensor.
Objetivo: Mantener la temperatura de la habitación cerca de la temperatura deseada.
Medida de desempeño (P): Proximidad a la temperatura objetivo (confort térmico) y ahorro de energía.

### 5.3. De PEAS a reglas de decisión

| Condición | Acción a tomar |
| :--- | :--- |
| Si la temperatura está muy por encima de la deseada | Encender aire acondicionado |
| Si la temperatura está muy por debajo de la deseada | Encender calefacción |
| Si la temperatura está cerca de la deseada | Mantener apagado |
```

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple (sin memoria, sin modelo del entorno).
    Percibe la temperatura actual del ambiente y decide si encender 
    la calefacción, el aire acondicionado, o no hacer nada, 
    buscando mantener la temperatura objetivo.
    """
    diferencia = temperatura_actual - temperatura_objetivo
    
    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefacción"
    else:
        return "Mantener apagado"
```

```python
# Secuencia original de la guía + 2 valores propios (21.0 límite inferior, 15.0 caso extremo frío)
percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 15.0]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepcion: {t}°C -> Accion del agente: {accion}")
```

```python
##Reflexión
¿Este agente termostato es racional?
Sí, es racional actuando como un agente reactivo simple. Toma la mejor decisión posible (la que maximiza su medida de desempeño) basándose únicamente en la percepción que recibe en ese preciso instante.

¿Bajo qué condiciones del entorno dejaría de serlo?
Dejaría de ser racional si:
1. El sensor de temperatura se malogra y envía datos erróneos.
2. Si una ventana se queda abierta; el agente encendería la calefacción infinitamente gastando energía sin lograr su objetivo, ya que no tiene "memoria" ni un modelo interno para entender que el calor se está escapando.
3. Si el margen configurado es demasiado amplio (ej. 5 grados), permitiendo que la temperatura sea incómoda antes de decidir actuar.
```
