## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Componente | Descripción |
| :--- | :--- |
| **Percepción (S)** | `tiempo_espera_minutos` (entero), `nivel_urgencia` ("baja", "media", "alta"), `cliente_premium` (booleano). |
| **Acciones (A)** | Escalar a nivel 1, escalar a nivel 2, escalar a nivel 3. |
| **Entorno (E)** | Sistema de gestión de incidencias y mesa de ayuda (Helpdesk). |
| **Objetivo** | Resolver incidentes eficientemente asignándolos al nivel de soporte adecuado según SLA. |
| **Medida de desempeño (P)** | Tiempo promedio de resolución de tickets y cumplimiento de Acuerdos de Nivel de Servicio (SLA). |

### Justificación de las reglas
Las reglas buscan priorizar adecuadamente los recursos del equipo de soporte:
Una urgencia "alta" o un cliente premium con espera excesiva (>15 min) requieren atención especializada inmediata en Nivel 3 para evitar penalizaciones contractuales.
La urgencia "media", los clientes premium o tiempos de espera prolongados (>30 min) se derivan a Nivel 2.
Las incidencias de urgencia "baja" sin prioridad especial se mantienen en Nivel 1 para no saturar a los especialistas.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta" or (cliente_premium and tiempo_espera_minutos > 15):
        return ("escalar a nivel 3", "Urgencia alta o cliente premium con tiempo de espera crítico (>15 min).")
    elif nivel_urgencia == "media" or cliente_premium or tiempo_espera_minutos > 30:
        return ("escalar a nivel 2", "Urgencia media, cliente premium o tiempo de espera acumulado (>30 min).")
    else:
        return ("escalar a nivel 1", "Incidencia de baja urgencia en cliente estándar sin espera prolongada.")
```

```python
# Pruebas con 6 combinaciones (incluyendo casos donde las condiciones compiten)
casos_soporte = [
    (10, "baja", False),   # Caso base -> Nivel 1
    (35, "baja", False),   # Urgencia baja pero espera alta (>30) -> Nivel 2
    (5, "baja", True),     # Urgencia baja pero cliente premium -> Nivel 2
    (20, "baja", True),    # Urgencia baja pero cliente premium + espera (>15) -> Nivel 3
    (10, "media", False),  # Urgencia media -> Nivel 2
    (2, "alta", False),    # Urgencia alta -> Nivel 3
]

print("--- RESULTADOS DE ESCALAMIENTO DE TICKETS ---")
for espera, urgencia, premium in casos_soporte:
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"Espera: {espera}m | Urgencia: {urgencia:<5} | VIP: {str(premium):<5} => [{accion.upper()}] - {motivo}")
```
